```python
import os
import json
from src import grava_csv, ler_json, list_to_text, move_file
```

```python
# variavel global 
# camada bronze
__RAWPENDINGPATH__ = '/Workspace/Users/claudiotarbe@gmail.com/databricks-lottery-pipeline/data/01_bronze/01_pending'
__RAWPROCESSEDPATH__ = '/Workspace/Users/claudiotarbe@gmail.com/databricks-lottery-pipeline/data/01_bronze/02_processed'

# camada silver
__SILVERPATH__ = '/Workspace/Users/claudiotarbe@gmail.com/databricks-lottery-pipeline/data/02_silver'
```

```python
for filename in os.listdir(__RAWPENDINGPATH__):

    localpath = f'{__RAWPENDINGPATH__}/{filename}'
    source = f'{__RAWPENDINGPATH__}/{filename}'
    destination = f'{__RAWPROCESSEDPATH__}/{filename}'
    
    if filename.endswith(".json"):

        data = ler_json(localpath)
        
        concurso = data.get('numero')
        
        dezenas = list_to_text(data.get('listaDezenas'),';')
        
        cabecalho = f'concurso;d1;d2;d3;d4;d5;d6\n'
        
        linha = f'{cabecalho}{concurso},{dezenas}'
        
        grava_csv(f'{__SILVERPATH__}/concurso_{concurso}.csv',linha)
        
        move_file(source, destination)

    else:
        
        print(f'{filename} não é um arquivo válido!')
```
